```python
path = r"/content/drive/My Drive/Case Comp/Training_final"

# Use glob to get all the CSV files in the folder
csv_files = glob.glob(os.path.join(path, "*.csv"))

# Loop over the list of CSV files
for f in csv_files:
    # Read the CSV file
    df = pd.read_csv(f)

    # Print the location and filename
    print('Location:', f)
    print('File Name:', f.split("/")[-1])  # Use forward slash for paths in Google Colab/Unix

    # Print the content
    display(df)
```

```python
folder_path = r'/content/drive/My Drive/Case Comp/Training_final'

# Read the CSV files using forward slashes for file paths
additional_features = pd.read_csv(f'{folder_path}/Additional Features.csv')
control_point = pd.read_csv(f'{folder_path}/Control Point.csv')
cost_utilization = pd.read_csv(f'{folder_path}/Cost & Utilization.csv')
demographics = pd.read_csv(f'{folder_path}/Demographics.csv')
pharmacy_utilization = pd.read_csv(f'{folder_path}/Pharmacy Utilization.csv')
sales_channel = pd.read_csv(f'{folder_path}/Sales Channel.csv')
social_determinants = pd.read_csv(f'{folder_path}/Social Determinants of Health.csv')
web_activity = pd.read_csv(f'{folder_path}/Web Activity.csv')
member_data = pd.read_csv(f'{folder_path}/MEMBER_DATA.csv')
quality_data = pd.read_csv(f'{folder_path}/QUALITY_DATA.csv')
member_details = pd.read_csv(f'{folder_path}/humana_mays_target_member_details.csv')
member_claim = pd.read_csv(f'{folder_path}/humana_mays_target_member_visit_claims.csv')
member_condition = pd.read_csv(f'{folder_path}/humana_mays_target_member_conditions.csv')
target_members = pd.read_csv(f'{folder_path}/humana_mays_target_members.csv')
```

```python
merged_df = target_members.merge(additional_features, on='id', how='left')
merged_df = merged_df.merge(control_point, on='id', how='left')
merged_df = merged_df.merge(cost_utilization, on='id', how='left')
merged_df = merged_df.merge(demographics, on='id', how='left')
merged_df = merged_df.merge(pharmacy_utilization, on='id', how='left')
merged_df = merged_df.merge(sales_channel, on='id', how='left')
merged_df = merged_df.merge(social_determinants, on='id', how='left')
merged_df = merged_df.merge(web_activity, on='id', how='left')
merged_df = merged_df.merge(member_data, on='id', how='left')
merged_df = merged_df.merge(member_claim, on='id', how='left')
merged_df= merged_df.merge(member_condition, on='id', how='left')
```
